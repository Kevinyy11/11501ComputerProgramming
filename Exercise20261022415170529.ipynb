## 二、題目說明（Problem Description）

請設計一個 Python 程式，建立學生分數分析系統。

程式必須能夠接收多筆分數，根據指定的倍率（multiplier）及加分值（bonus）調整分數，接著計算調整後的分數平方、總分及平均分數，最後回傳分析結果。

### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 2：呼叫函數

已知下列資料：

```python
scores = [60, 70, 80]

settings = {
    "multiplier": 1.1,
    "bonus": 5
}
```

**問題 1：** 請使用 `*` 與 `**` 解包 `scores` 和 `settings`，呼叫 `analyze_scores()` 函數，並將結果存入變數 `result`。

```python
result = analyze_scores(*scores, **settings)
```

### Part 3：思考與分析（Questions）

1. 在函數內部，`scores` 的資料型態是什麼？其值為何？
資料型態：tuple（元組）。

其值：(60, 70, 80)。

原因：在函數定義中使用 *scores，會將所有傳入的多個位置參數打包成一個元組。

2. 在函數內部，`options` 的資料型態是什麼？其值為何？
資料型態：dict（字典）。

其值：{"multiplier": 1.1, "bonus": 5}。

原因：在函數定義中使用 **options，會將所有傳入的關鍵字參數打包成一個字典。

3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。
*：在此處作為算術運算子代表乘法（將調整前的分數乘上倍率）。

**：在此處作為算術運算子代表次方（將調整後的分數計算平方，即二次方）。

4. 為什麼函數使用 `return`，而不是只使用 `print()`？
return 會將計算結果回傳給呼叫端，使該資料可以被指派給變數（如 result），並在後續的程式邏輯中重複利用、儲存或傳遞。

若僅使用 print()，雖然能將結果印在螢幕上，但函數本身的回傳值會是 None，無法在後續程式中繼續進行運算。

5. 如果呼叫 `analyze_scores(60, 70, 80)`，但沒有提供 `settings`，程式會如何執行？請說明原因。
程式可以正常執行。

原因：因為程式碼使用了 options.get("multiplier", 1) 與 options.get("bonus", 0)。當字典中找不到對應的鍵時，get() 方法會自動採用預設值（multiplier = 1，bonus = 0），而不會引發 KeyError。

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

## 四、總結

請根據本題，說明 `*`、`**` 與 `return` 在 Python 中的不同用途，並舉出至少一個程式碼例子，展示同一個符號在不同語法位置所代表的意義。
* 的三種主要用途：

算術運算：代表乘法（例如 score * multiplier）。

函數定義（*args）：將不定數量的位置參數打包收集成一個 tuple。

函數呼叫（*iterable）：將可迭代物件（如串列、元組）解包成獨立的位置參數。

** 的三種主要用途：

算術運算：代表次方（例如 score ** 2）。

函數定義（**kwargs）：將不定數量的關鍵字參數打包收集成一個 dict。

函數呼叫（**dict）：將字典解包成獨立的關鍵字參數。

return 的用途：

用於終止函數的執行，並將計算結果回傳給呼叫端，使資料得以在程式其他地方繼續被處理與運用。
